# 04 — Complete three-task engineering pilot

This notebook reviews only the privacy-safe aggregate pilot report. It does not open patient-level texts, indices, embeddings, labels, predictions, or models.

## Frozen pilot scope and final-panel isolation

```python
import json
from pathlib import Path

report_path = Path('reports/generated/pilot/three_task_m1_engineering_pilot_v1/pilot_results.json')
report = json.loads(report_path.read_text(encoding='utf-8'))
print(json.dumps(report['scientific_scope'], indent=2, sort_keys=True))
print('Final panel used for pilot:', report['final_panel_used_for_pilot'])
```

## Train/validation cohort and preparation coverage

```python
print(json.dumps(report['prepare']['cohort_groups'], indent=2, sort_keys=True))
for task in report['prepare']['task_summaries']:
    print(task['task'], task['source_instances'], task['eligible_text_rows'], task['content_matching_passed'])
```

## Validation metrics, calibration, and paired contrasts

```python
for task, task_report in report['tasks'].items():
    print('TASK', task)
    print('B4', json.dumps(task_report['B4_training_prevalence'], sort_keys=True))
    print('B3', json.dumps(task_report['B3_governed_code_count']['metrics'], sort_keys=True))
    print('contrasts', json.dumps(task_report['paired_contrasts'], sort_keys=True))
    print('C3', json.dumps(task_report['c3_seed_summary'], sort_keys=True))
```

## Token coverage, throughput, failures, and retries

```python
print(json.dumps(report['throughput'], indent=2, sort_keys=True))
print(json.dumps(report['embedding_tasks'], indent=2, sort_keys=True))
```

```python
checklist = {
    'pilot_status_pass': report['status'] == 'PASS',
    'content_matching_passed': report['content_matching_passed'],
    'final_panel_unchanged': not report['final_panel_used_for_pilot'],
    'test_predictions_or_performance_used': report['test_predictions_or_performance_used'],
    'm2_run': report['m2_run'],
    'm3_run': report['m3_run'],
    'qwen_8b_run': report['qwen_8b_run'],
}
print(json.dumps(checklist, sort_keys=True))
```
