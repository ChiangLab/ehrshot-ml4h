# 05 — Locked four-task final foundation review

This thin notebook reads only the privacy-safe aggregate foundation report. It never opens the final cohort, histories, prediction times, labels, patient-window rows, or prepared texts.

```python
import json
from pathlib import Path

path = Path('reports/generated/final/locked_four_task_m1_v1/foundation.json')
report = json.loads(path.read_text(encoding='utf-8'))
assert report['status'] == 'PASS'
assert report['test_predictions_or_performance_used'] is False
print(report['protocol_decision'])
print(json.dumps(report['scientific_scope'], indent=2, sort_keys=True))
```

## Patient-balanced cohort and official split coverage

```python
for group in report['cohort']['groups']:
    print(group['task'], group['split'], group['instances'], round(group['prevalence'], 4))
assert report['cohort']['maximum_instances_per_patient_task'] == 1
assert report['cohort']['test_instances_selected'] == 0
```

## Outcome definitions, shortcut exclusion, and serializer gates

```python
for task, audit in report['outcome_and_shortcut_audits'].items():
    print(task, audit['status'], audit['outcome_horizon'])
    assert audit['target_defining_future_events_retained'] == 0
for summary in report['task_summaries']:
    print(
        summary['task'], 'instances=', summary['source_instances'],
        'metadata=', summary['patient_window_metadata_rows'],
    )
    assert summary['content_matching_passed']
    assert summary['metadata_mirrors_every_embedding_row']
    assert summary['clinical_event_arrays_rearranged_by_c2_or_c3'] is False
```

## Boundary

This foundation authorizes final M1 throughput and embedding only. M2, M3, and held-out evaluation remain prohibited.
