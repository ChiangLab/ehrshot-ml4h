# 06 — Final four-task M1 validation

This thin notebook reads only the reviewed aggregate report. It never opens labels, predictions, identifiers, timestamps, histories, embeddings, or private cache metadata.

```python
import json
from pathlib import Path

path = Path('reports/generated/final/locked_four_task_m1_v1/final_m1_validation.json')
report = json.loads(path.read_text(encoding='utf-8'))
assert report['status'] == 'PASS'
assert report['test_predictions_or_performance_used'] is False
assert report['test_results_reported'] is False
print(report['protocol_decision'])
print(json.dumps(report['scientific_scope'], indent=2, sort_keys=True))
```

## Equal-weight panel validation estimates

```python
for name, result in report['equal_weight_panel_macro'].items():
    metrics = result['metrics']
    print(
        name, 'AUROC=', round(metrics['auroc'], 4),
        'AUPRC=', round(metrics['auprc'], 4),
        'Brier=', round(metrics['brier_score'], 4),
    )
```

## Prespecified temporal and baseline contrasts

```python
for item in report['equal_weight_panel_macro_paired_bootstrap_contrasts']:
    print(
        item['first'], '-', item['second'], '=', round(item['estimate'], 4),
        '95% CI', [round(value, 4) for value in item['ci_95']],
    )
for task, result in report['tasks'].items():
    variability = result['c3_seed_variability']
    print(task, 'C3 SD=', round(variability['standard_deviation'], 6))
```

## Runtime-only 8B feasibility and operational evidence

```python
plan = report['throughput']['feasibility_plan']
assert plan['outcome_performance_used_for_feasibility'] is False
for task, decision in plan['tasks'].items():
    print(task, decision['confirmatory_priorities_selected'])
print('Runtime seconds:', report['runtime_seconds'])
print('Private storage bytes:', report['private_storage_bytes'])
```
